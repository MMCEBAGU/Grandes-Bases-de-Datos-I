```python
# tratamiento de datos 
import numpy as np
import pandas as pd

# graficos
import matplotlib.pyplot as plt

# preprocesado y modelado
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

from numpy import linalg as LA
```

```python
import pandas as pd
df = pd.read_csv("iris.csv")
df.head()
```

```python
df.shape
```

```python
# Determinar el número de renglones de la base de datos
index = df.index
renglones = len(index)

renglones
```

```python
# adaptacion a nuestra vase para quitar la variable no numerica
df_pca = df.iloc[:, 0:4]
df_pca
```

```python
# Estandarización de las variables
df2 = StandardScaler().fit_transform(df_pca)
df2
```

```python
# Cálculo de la matriz de correlaciones para la matriz transformada
A = (1/renglones) * np.dot(df2.T, df2)
A
```

```python
# Entrenamiento del modelo PCA con el escalado de los datos
pca_pipe = make_pipeline(StandardScaler(), PCA())
pca_pipe.fit(df_pca)

# Se extrae el modelo entrenado del pipeline
modelo_pca = pca_pipe.named_steps["pca"]

print("Eigenvalores:")
eigenvalores = LA.eigvals(A)
print(eigenvalores)
```

```python
# Porcentaje de la varianza explicada por cada nuevo componente
print("porcentaje de varianza explicada por componente")
print(modelo_pca.explained_variance_ratio_)
```

```python
# Calculo de los eigenvectores
print("Eigenvectores (por renglón):")

pd.DataFrame(data=modelo_pca.components_,
             columns=df_pca.columns,
             index=["PC1", "PC2", "PC3", "PC4"])
```

```python
# proyecciones de los componentes
proyecciones = np.dot(modelo_pca.components_, df2.T)
proyecciones = pd.DataFrame(proyecciones,
                            index = ["PC1","PC2","PC3","PC4"])
proyecciones = proyecciones.transpose().set_index(df.index)
proyecciones
```

```python
x = proyecciones.iloc[:,0]
y = proyecciones.iloc[:,1]
z = df.index
x = x.to_numpy()
y = y.to_numpy()

fig, ax = plt.subplots()
ax.set_title("Mapa de Observaciones")
ax.scatter(x,y)

for i, txt in enumerate(z):
    ax.annotate(txt,(x[i],y[i]))
```

```python
componentes_2 = pd.DataFrame(data=modelo_pca.components_,
                             columns=df_pca.columns,
                             index=["PC1", "PC2", "PC3", "PC4"])

componentes_2 = componentes_2.iloc[0:2, :]
componentes_2 = componentes_2.T
componentes_2
```

```python
x = componentes_2.iloc[:, 0]
y = componentes_2.iloc[:, 1]
z = componentes_2.index

x = x.to_numpy()
y = y.to_numpy()

fig, ax = plt.subplots()

ax.set_title("Mapa de Factores")

ax.scatter(x, y)

for i, txt in enumerate(z):
    ax.annotate(txt, (x[i], y[i]))

plt.show()
```

```python
# En el mapa se puede observar que las variables sepal.length, petal.width
# petal.legth tienen mayor relacion con el primer componente (PC1) en cambio 
# la variable sepal.width tiene mayor relacion con el segundo componente (PC2)
```

```python
# Procedimiento para obtener la matriz estandarizada original
original = np.dot(modelo_pca.components_.T, proyecciones.T)
original = original.T
print(original)
```
